INTEGRASI DATA

```python
import pandas as pd

# Membaca semua isi file dataset
files = [
    "Jumlah Penumpang Pesawat (Angkutan Udara) Domestik di 5 Bandara Utama, 2020.csv",
    "Jumlah Penumpang Pesawat (Angkutan Udara) Domestik di 5 Bandara Utama, 2021.csv",
    "Jumlah Penumpang Pesawat (Angkutan Udara) Domestik di 5 Bandara Utama, 2022.csv",
    "Jumlah Penumpang Pesawat (Angkutan Udara) Domestik di 5 Bandara Utama, 2023.csv",
    "Jumlah Penumpang Pesawat (Angkutan Udara) Domestik di 5 Bandara Utama, 2024.csv",
    "Jumlah Penumpang Pesawat (Angkutan Udara) Domestik di 5 Bandara Utama, 2025.csv"
]

dfs = []

for file in files:
    # Membaca semua baris isi file dataset
    df = pd.read_csv(file, header=None)

    # Mengubah semua header "Unnamed" menjadi string kosong
    df.columns = [
        "" if str(c).startswith("Unnamed") else str(c)
        for c in df.columns
    ]

    # Menghapus baris yang kosong
    df = df.dropna(how="all")

    dfs.append(df)

# Concatenate (Menggabungkan) seluruh dataframe
df_all = pd.concat(dfs, ignore_index=True)

# Menyimpan file integrasi
output_path = "Hasil_Integrasi.csv"
df_all.to_csv(output_path, index=False, header=True, encoding="utf-8-sig")

print("File tersimpan sebagai:", output_path)
print("Jumlah baris:", len(df_all))
print("Jumlah kolom:", len(df_all.columns))
```

CLEANING HASIL INTEGRASI

```python
import pandas as pd
import numpy as np
import os
import re

input_file = "Hasil_Integrasi.csv"
output_file = "Dataset_Penumpang_Clean.csv"

# Nama bulan
MONTHS_ID = ["Januari","Februari","Maret","April","Mei","Juni",
             "Juli","Agustus","September","Oktober","November","Desember"]

# Membersihkan spasi, koma, titik, (-)
def clean_num(x):
    if pd.isna(x) or str(x).strip() == "":
        return 0
    s = str(x).strip().replace(",", "").replace(".", "").replace(" ", "")
    if re.fullmatch(r'[-–—]+', s):
        return 0
    return int(re.sub(r'[^\d\-]', '', s) or 0)

# Parsing file hasil integrasi
def parse_integrated_file(path):
    df_raw = pd.read_csv(path, header=None, dtype=str, keep_default_na=False)
    nrows, ncols = df_raw.shape
    all_records = []

    # Blok kolom tiap kategori
    # Tiap blok terdiri dari 13 kolom yaitu 12 bulan + tahunan
    category_col_blocks = {
        "Keberangkatan": (1, 13),
        "Kedatangan": (14, 26),
        "Transit": (27, 39)
    }

    r_idx = 0
    while r_idx < nrows:
        col1_val = str(df_raw.iat[r_idx, 1]).strip()
        if col1_val.isdigit() and len(col1_val) == 4:
            tahun = int(col1_val)
            data_start = r_idx + 2
            data_end = min(data_start + 6, nrows)

            for dr in range(data_start, data_end):
                airport = str(df_raw.iat[dr, 0]).strip()
                if not airport or airport.lower() in ["nan","none","total","jumlah"]:
                    continue

                # ambil data tiap kategori
                for cat, (start_col, end_col) in category_col_blocks.items():
                    rec = {"bandara": airport, "kategori": cat, "tahun": tahun}
                    for m in MONTHS_ID + ["Tahunan"]:
                        rec[m] = 0

                    month_counter = 0
                    for c in range(start_col, end_col+1):
                        if c < ncols:
                            month_name = None
                            if month_counter < len(MONTHS_ID):
                                month_name = MONTHS_ID[month_counter]
                            elif month_counter == len(MONTHS_ID):
                                month_name = "Tahunan"
                            if month_name:
                                rec[month_name] = clean_num(df_raw.iat[dr, c])
                                month_counter += 1
                        else:
                            break
                    all_records.append(rec)
            r_idx += 12
            continue
        r_idx += 1

    if not all_records:
        print("Ga ada data yang ke-parse, cek file atau logikanya")
        return pd.DataFrame()

    df_parsed = pd.DataFrame(all_records)
    cols_order = ["bandara","kategori","tahun"] + MONTHS_ID + ["Tahunan"]
    for c in cols_order:
        if c not in df_parsed.columns:
            df_parsed[c] = 0
    df_parsed = df_parsed[cols_order]

    # Normalisasi nama bandara
    df_parsed["bandara"] = df_parsed["bandara"].astype(str).str.replace(r"(?i)\s+Medan$", "", regex=True).str.strip()
    df_parsed["bandara"] = df_parsed["bandara"].replace({
        r"(?i)soekarno[\s\-]?hatta|soetta": "Soekarno-Hatta",
        r"(?i)juanda": "Juanda",
        r"(?i)kualanamu": "Kualanamu",
        r"(?i)sultan\s+hasanuddin": "Sultan Hasanuddin",
        r"(?i)(i\s+gusti\s+ngurah\s+rai|ngurah\s+rai|ngurahrai)": "Ngurah Rai"
    }, regex=True)
    return df_parsed

# Perhitungan statistics
df_raw_integrated = pd.read_csv(input_file, header=None)
missing_before = df_raw_integrated.isna().sum()
duplicates_before = df_raw_integrated.duplicated().sum()

df_clean = parse_integrated_file(input_file)

if not df_clean.empty:
    missing_after = df_clean.isna().sum()
    duplicates_after = df_clean.duplicated().sum()

    # Cek outliers per bulan
    outlier_after = {}
    for m in MONTHS_ID:
        if m in df_clean.columns:
            col = df_clean[m]
            mean, std = col.mean(), col.std()
            outlier_after[m] = col[(col > mean + 3*std) | (col < mean - 3*std)].count()

    bandara_unique_before = df_raw_integrated[0].nunique()
    kategori_unique_before = df_raw_integrated[1].nunique()
    bandara_unique_after = df_clean['bandara'].nunique()
    kategori_unique_after = df_clean['kategori'].nunique()

    os.makedirs(os.path.dirname(output_file) or ".", exist_ok=True)
    df_clean.to_csv(output_file, index=False)


    print(" Missing Values ")
    print("Sebelum:", missing_before)
    print("Sesudah:", missing_after)

    print("\n Duplicates ")
    print("Sebelum:", duplicates_before)
    print("Sesudah:", duplicates_after)

    print("\n Outliers per bulan ")
    for m, count in outlier_after.items():
        print(f"{m}: {count}")

    print("\n Bandara atau Kategori unik ")
    print("Bandara unik sebelum:", bandara_unique_before)
    print("Kategori unik sebelum:", kategori_unique_before)
    print("Bandara unik setelah:", bandara_unique_after)
    print("Kategori unik setelah:", kategori_unique_after)

    print("\nFile hasil cleaning:", output_file)
    print("Ukuran df:", df_clean.shape)
    print(df_clean.head())
else:
    print("Error: df_clean kosong, cek input atau logikanya")
    print("Ukuran df: (0,0)")
```

VISUALISASI HASIL CLEANING

```python

df = pd.read_csv("Dataset_Penumpang_Clean.csv")

import pandas as pd
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets
from IPython.display import display

# Menggunakan Dataset
df = df_clean.copy()

# Memastikan kolom bulanan menjadi numerik
for col in MONTHS_ID + ["Tahunan"]:
    df[col] = df[col].apply(clean_num)

# Plotting
def plot_bandara(selected_tahun, selected_bandara, selected_kategori):
    df_plot = df[
        (df['tahun'] == selected_tahun) &
        (df['bandara'] == selected_bandara) &
        (df['kategori'] == selected_kategori)
    ]

    if df_plot.empty:
        print("Tidak ada data untuk filter yang dipilih.")
        return

    y_values = df_plot[MONTHS_ID].values.flatten()

    plt.figure(figsize=(10,5))
    plt.plot(MONTHS_ID, y_values, marker='o', linestyle='-', color='b')
    plt.title(f"Jumlah Penumpang per Bulan\n{selected_bandara} - {selected_kategori} ({selected_tahun})")
    plt.xlabel("Bulan")
    plt.ylabel("Jumlah Penumpang")
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.show()

# Widgets
tahun_widget = widgets.Dropdown(
    options=sorted(df['tahun'].unique()),
    description='Tahun:'
)

bandara_widget = widgets.Dropdown(
    options=sorted(df['bandara'].unique()),
    description='Bandara:'
)

kategori_widget = widgets.Dropdown(
    options=sorted(df['kategori'].unique()),
    description='Kategori:'
)

interact(plot_bandara, selected_tahun=tahun_widget, selected_bandara=bandara_widget, selected_kategori=kategori_widget)
```
